# 7. References and Resources

A curated collection of textbooks, videos, and interactive tools for learning linear algebra.  
Organised by topic so you can dive deeper into whatever interests you most.

## Textbooks

| Resource | Notes |
|----------|-------|
| **Gilbert Strang, *Introduction to Linear Algebra*, 5th ed.** (Wellesley-Cambridge Press, 2016) | The classic. Clear exposition with good examples. Covers everything from vectors through SVD. |
| [Interactive Linear Algebra (Georgia Tech)](https://textbooks.math.gatech.edu/ila/index.html) | Free online textbook with embedded interactive figures. Great for geometric intuition. |
| [The Art of Linear Algebra (GitHub)](https://github.com/kenjihiranabe/The-Art-of-Linear-Algebra) | Visual summary of Strang's key concepts -- one-page graphic maps of the main ideas. |

## Video Courses

| Resource | Notes |
|----------|-------|
| [3Blue1Brown -- Essence of Linear Algebra](https://www.youtube.com/playlist?list=PLZHQObOWTQDPD3MizzM2xVFitgF8hE_ab) | Outstanding animated series that builds geometric intuition. Covers transformations, determinants, eigenvalues, and more. Highly recommended as a first pass. |
| [MIT OCW 18.06 -- Linear Algebra (Gilbert Strang)](https://ocw.mit.edu/courses/18-06-linear-algebra-spring-2010/) | Full university course with video lectures, problem sets, and exams. |
| [Khan Academy -- Linear Algebra](https://www.khanacademy.org/math/linear-algebra) | Comprehensive, beginner-friendly video lessons with practice exercises. |

## Interactive Tools

| Resource | Notes |
|----------|-------|
| [GeoGebra Matrix Calculator](https://www.geogebra.org/m/eTfDpMVa) | Visual matrix transformations in the browser. |
| [Immersive Linear Algebra](http://immersivemath.com/ila/index.html) | Free interactive online book with 3D figures you can rotate and manipulate. |
| [Wolfram Alpha](https://www.wolframalpha.com/) | Quick computation of eigenvalues, SVD, determinants, etc. |
