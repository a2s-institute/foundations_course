# 1. Vectors and Linear Combinations

This notebook builds geometric intuition for vectors and their operations in 2D and 3D.
By the end you should be able to *see* what linear combinations, span, basis, and linear independence mean -- not just compute them.

**Prerequisites:** Basic coordinate geometry (plotting points on a grid).

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

%matplotlib inline

COLORS = {"v1": "#2196F3", "v2": "#F44336", "result": "#4CAF50", "grid": "#E0E0E0"}

def setup_2d_axes(ax, lim=5, title=""):
    """Configure a 2D axes with grid, equal aspect, and centered origin."""
    ax.set_xlim(-lim, lim)
    ax.set_ylim(-lim, lim)
    ax.set_aspect("equal")
    ax.grid(True, linewidth=0.5, alpha=0.5)
    ax.axhline(0, color="k", linewidth=0.6)
    ax.axvline(0, color="k", linewidth=0.6)
    if title:
        ax.set_title(title, fontsize=13)

def draw_vector(ax, v, origin=(0, 0), color="blue", label=None, lw=2):
    """Draw a 2D vector arrow from *origin*."""
    ax.annotate(
        "", xy=(origin[0] + v[0], origin[1] + v[1]), xytext=origin,
        arrowprops=dict(arrowstyle="->", color=color, lw=lw),
    )
    if label:
        mid = (origin[0] + v[0] / 2, origin[1] + v[1] / 2)
        ax.text(mid[0] + 0.15, mid[1] + 0.15, label, fontsize=11, color=color)
```

## 1.1 What Is a Vector?

A **vector** is an arrow with a *direction* and a *magnitude* (length).  
In 2D we write $\mathbf{v} = \begin{pmatrix} v_x \\ v_y \end{pmatrix}$, where $v_x$ and $v_y$ tell us how far to go along each axis.

Crucially, two arrows that point the same way and have the same length represent the **same** vector, no matter where they start.

```python
fig, ax = plt.subplots(figsize=(6, 6))
setup_2d_axes(ax, title="The same vector drawn from different origins")

v = np.array([3, 2])
origins = [(0, 0), (-4, -3), (1, -2), (-2, 2)]
for i, o in enumerate(origins):
    draw_vector(ax, v, origin=o, color=COLORS["v1"],
                label="v" if i == 0 else None)

ax.set_xlabel("x")
ax.set_ylabel("y")
plt.tight_layout()
plt.show()
```

## 1.2 Vector Addition -- The Parallelogram Rule

To add two vectors $\mathbf{a}$ and $\mathbf{b}$, place the tail of $\mathbf{b}$ at the tip of $\mathbf{a}$.  
The result $\mathbf{a} + \mathbf{b}$ goes from the tail of $\mathbf{a}$ to the tip of $\mathbf{b}$.  
This traces out a **parallelogram**.

```python
def plot_addition(a, b):
    fig, ax = plt.subplots(figsize=(6, 6))
    setup_2d_axes(ax, title="Vector addition: a + b")

    draw_vector(ax, a, color=COLORS["v1"], label="a", lw=2.5)
    draw_vector(ax, b, color=COLORS["v2"], label="b", lw=2.5)
    draw_vector(ax, b, origin=tuple(a), color=COLORS["v2"], lw=1.5)
    draw_vector(ax, a, origin=tuple(b), color=COLORS["v1"], lw=1.5)
    draw_vector(ax, a + b, color=COLORS["result"], label="a+b", lw=2.5)

    ax.fill([0, a[0], a[0]+b[0], b[0]], [0, a[1], a[1]+b[1], b[1]],
            alpha=0.08, color=COLORS["result"])
    plt.tight_layout()
    plt.show()

a = np.array([-3, -1])
b = np.array([1, 3])
plot_addition(a, b)
```

## 1.3 Scalar Multiplication

Multiplying a vector by a scalar $c$ **stretches** it (if $|c|>1$), **shrinks** it (if $|c|<1$), or **flips** it (if $c<0$).

$$c \, \mathbf{v} = \begin{pmatrix} c \, v_x \\ c \, v_y \end{pmatrix}$$

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
v = np.array([2, 1])
scalars = [1, 2, 0.5, -1]
titles = ["c = 1 (original)", "c = 2 (stretch)", "c = 0.5 (shrink)", "c = -1 (flip)"]

for ax, c, t in zip(axes, scalars, titles):
    setup_2d_axes(ax, lim=5, title=t)
    draw_vector(ax, v, color=COLORS["grid"], label="v", lw=1.5)
    draw_vector(ax, c * v, color=COLORS["v1"], label=f"{c}v", lw=2.5)

plt.tight_layout()
plt.show()
```

## 1.4 Linear Combinations (Interactive)

A **linear combination** of vectors $\mathbf{v}_1$ and $\mathbf{v}_2$ is

$$c_1 \mathbf{v}_1 + c_2 \mathbf{v}_2$$

Use the sliders below to explore how changing $c_1$ and $c_2$ sweeps out different points in the plane.

```python
v1 = np.array([2, 0.5])
v2 = np.array([0.5, 2])

def plot_linear_combination(c1, c2):
    fig, ax = plt.subplots(figsize=(6, 6))
    setup_2d_axes(ax, title=f"Linear combination: {c1:.1f} v1 + {c2:.1f} v2")

    # Dashed reference arrows for the original (unscaled) basis vectors
    for v, col, lbl in [(v1, COLORS["v1"], "v1"), (v2, COLORS["v2"], "v2")]:
        ax.annotate("", xy=v, xytext=(0, 0),
                    arrowprops=dict(arrowstyle="->", color=col, lw=1, linestyle="--"))
        ax.text(v[0] + 0.12, v[1] + 0.12, lbl, fontsize=9, color=col, alpha=0.55)

    scaled_v1 = c1 * v1
    scaled_v2 = c2 * v2
    result = scaled_v1 + scaled_v2

    draw_vector(ax, scaled_v1, color=COLORS["v1"], lw=2, label=f"{c1:.1f}v1")
    draw_vector(ax, scaled_v2, origin=tuple(scaled_v1), color=COLORS["v2"], lw=2, label=f"{c2:.1f}v2")
    draw_vector(ax, result, color=COLORS["result"], lw=2.5, label="result")

    ax.plot(*result, "o", color=COLORS["result"], markersize=7)
    plt.tight_layout()
    plt.show()

widgets.interact(
    plot_linear_combination,
    c1=widgets.FloatSlider(value=1, min=-3, max=3, step=0.1, description="c1"),
    c2=widgets.FloatSlider(value=1, min=-3, max=3, step=0.1, description="c2"),
);
```

## 1.5 Span

The **span** of a set of vectors is the collection of *all* their linear combinations.

- Two non-parallel vectors in 2D span the entire plane.
- Two parallel vectors only span a line.
- A single non-zero vector spans a line through the origin.

Below we sample many random coefficients and plot the resulting points to visualise the span.

```python
rng = np.random.default_rng(42)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Case 1: non-parallel vectors -> span is the whole plane
ax = axes[0]
setup_2d_axes(ax, title="Non-parallel vectors: span = entire plane")
u1, u2 = np.array([2, 0.5]), np.array([0.5, 2])
coeffs = rng.uniform(-4, 4, size=(4000, 2))
pts = coeffs[:, 0:1] * u1 + coeffs[:, 1:2] * u2
ax.scatter(pts[:, 0], pts[:, 1], s=4, alpha=0.4, color=COLORS["result"])
draw_vector(ax, u1, color=COLORS["v1"], label="v1", lw=2.5)
draw_vector(ax, u2, color=COLORS["v2"], label="v2", lw=2.5)

# Case 2: parallel vectors -> span is a line
ax = axes[1]
setup_2d_axes(ax, title="Parallel vectors: span = a line")
u1, u2 = np.array([2, 1]), np.array([4, 2])
pts = coeffs[:, 0:1] * u1 + coeffs[:, 1:2] * u2
ax.scatter(pts[:, 0], pts[:, 1], s=4, alpha=0.4, color=COLORS["result"])
draw_vector(ax, u1, color=COLORS["v1"], label="v1", lw=2.5)
draw_vector(ax, u2, color=COLORS["v2"], label="v2", lw=2.5)

plt.tight_layout()
plt.show()
```

## 1.6 Basis and Linear Independence

A set of vectors is **linearly independent** if none of them can be written as a linear combination of the others.  
A **basis** for a space is a *linearly independent set* that **spans** the whole space.

In 2D, any two non-parallel vectors form a basis.
The standard basis is $\mathbf{e}_1 = (1,0)^T$ and $\mathbf{e}_2 = (0,1)^T$.

**Geometric test:** two 2D vectors are linearly *dependent* exactly when they are parallel (one is a scalar multiple of the other).

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ax = axes[0]
setup_2d_axes(ax, lim=4, title="Linearly independent (a basis)")
draw_vector(ax, np.array([2, 1]), color=COLORS["v1"], label="v1", lw=2.5)
draw_vector(ax, np.array([1, 3]), color=COLORS["v2"], label="v2", lw=2.5)

ax = axes[1]
setup_2d_axes(ax, lim=4, title="Linearly dependent (NOT a basis)")
draw_vector(ax, np.array([2, 1]), color=COLORS["v1"], label="v1", lw=2.5)
draw_vector(ax, np.array([4, 2]), color=COLORS["v2"], label="v2 = 2·v1", lw=2.5)

plt.tight_layout()
plt.show()
```

## 1.7 Extending to 3D

Everything above generalises naturally.  Three linearly independent vectors in 3D span all of $\mathbb{R}^3$.  
Below is a quick 3D visualisation.

```python
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot(111, projection="3d")

origin = np.zeros(3)
vecs = [np.array([2, 0, 0]), np.array([0, 2, 0]), np.array([0, 0, 2])]
colors = [COLORS["v1"], COLORS["v2"], COLORS["result"]]
labels = ["e1", "e2", "e3"]

for v, c, lbl in zip(vecs, colors, labels):
    ax.quiver(*origin, *v, color=c, arrow_length_ratio=0.12, linewidth=2.5, label=lbl)

ax.set_xlim([-1, 3]); ax.set_ylim([-1, 3]); ax.set_zlim([-1, 3])
ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
ax.set_title("Standard basis in 3D")
ax.legend()
plt.tight_layout()
plt.show()
```

### Exercise 1

Are the vectors $\mathbf{a} = (1, 3)^T$ and $\mathbf{b} = (2, 6)^T$ linearly independent?  
Justify your answer geometrically and algebraically.

<details>
<summary><b>Solution</b></summary>

No. $\mathbf{b} = 2\mathbf{a}$, so they are parallel and therefore linearly **dependent**.  
Algebraically, the determinant of $[\mathbf{a} \;|\; \mathbf{b}]$ is $1 \cdot 6 - 3 \cdot 2 = 0$.

### Exercise 2

Express the vector $\mathbf{w} = (7, 4)^T$ as a linear combination of $\mathbf{v}_1 = (2, 1)^T$ and $\mathbf{v}_2 = (1, 2)^T$.  
Find $c_1, c_2$ such that $c_1 \mathbf{v}_1 + c_2 \mathbf{v}_2 = \mathbf{w}$.

<details>
<summary><b>Solution</b></summary>

Solve the system:

$$2c_1 + c_2 = 7$$
$$c_1 + 2c_2 = 4$$

From the first equation: $c_2 = 7 - 2c_1$.  
Substituting: $c_1 + 2(7 - 2c_1) = 4 \Rightarrow -3c_1 = -10 \Rightarrow c_1 = 10/3$.  
$c_2 = 7 - 20/3 = 1/3$.  

Check: $(10/3)(2,1)^T + (1/3)(1,2)^T = (20/3 + 1/3,\; 10/3 + 2/3)^T = (7, 4)^T$ ✓

</details>

### Exercise 3

Use the interactive widget in Section 1.4 to find (approximately) the coefficients $c_1, c_2$ that place the result vector at the point $(3, 3)$.  
Then verify your answer by computing $c_1 \mathbf{v}_1 + c_2 \mathbf{v}_2$ with numpy.

<details>
<summary><b>Solution</b></summary>

With $\mathbf{v}_1 = (2, 0.5)^T$ and $\mathbf{v}_2 = (0.5, 2)^T$:

$$2c_1 + 0.5c_2 = 3$$
$$0.5c_1 + 2c_2 = 3$$

$c_1 = c_2 = 6/5 = 1.2$ (by symmetry of the system).

</details>

```python
# Verification cell for Exercise 3
v1 = np.array([2, 0.5])
v2 = np.array([0.5, 2])
target = np.array([3, 3])

A = np.column_stack([v1, v2])
coeffs = np.linalg.solve(A, target)
print(f"c1 = {coeffs[0]:.4f}, c2 = {coeffs[1]:.4f}")
print(f"Check: c1*v1 + c2*v2 = {coeffs[0]*v1 + coeffs[1]*v2}")
```
