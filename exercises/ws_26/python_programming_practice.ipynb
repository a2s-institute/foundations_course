**Task:** Use a `while` loop to print the squares of the first 8 natural numbers and accumulate their sum.

**Task:** Write a `while` loop that determines whether a given number is prime.

```python
num = 37
is_prime = True

if num > 1:
    i = 2
    while i < num:
        if num % i == 0:
            is_prime = False
            break
        i += 1

print(f"{num} is {'a prime' if is_prime else 'not a prime'} number")
```

Note: a `for...else` construct runs the `else` block only if the loop completes without hitting a `break`.

```python
num = 37
if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print(f"{num} is not a prime number")
            break
    else:
        print(f"{num} is a prime number")
```

**Task:** Write a function that takes two lists, reverses the first one *in place*, and returns the sum of both lists multiplied together.

```python
def combine_lists(list_a, list_b):
    list_a.reverse()
    return sum(list_a) * sum(list_b)

original = [1, 2, 4]
print(combine_lists(original, [1, 1]))
print(original)   # notice the list was reversed in place
```

**Task:** Write a recursive function to compute the factorial of a number.

```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

factorial(6)
```

## 12. Object-Oriented Programming

A **class** is a blueprint; an **object** is an instance of that class.

**Task:** Create a class `LabMember` with attributes `name`, `age`, and `role`. Instantiate three objects and print their details.

```python
class LabMember:
    def __init__(self, name, age, role):
        self.name = name
        self.age = age
        self.role = role

    def print_details(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Role: {self.role}")

    def compute_stipend(self):
        # return 1500 if self.age < 30 else 2000
        if self.age:
            return 1500
        else:
            return 2000

member1 = LabMember("Anja", 27, "PhD Candidate")
member2 = LabMember("Marco", 34, "Postdoc")

member1.print_details()
print(f"Stipend: {member1.compute_stipend()}")
member2.print_details()
```

### Inheritance
**Task:** Create a subclass `SeniorMember` that inherits from `LabMember` and overrides `compute_stipend` to add a bonus.

```python
class SeniorMember(LabMember):
    def __init__(self, name, age, role, exp):
        super().__init__(name, age, role)
        self.exp = exp

    def compute_stipend(self, bonus=500):
        base = super().compute_stipend()
        return base + bonus
    
    def return_exp(self):
        return f"{self.exp} year of experience"

senior = SeniorMember("Elena", 45, "Group Leader", 3)
print(senior.compute_stipend())
print(senior.return_exp())

# print(senior.__dict__)
# print(help(senior))
```

### Encapsulation, Polymorphism & Abstraction

- **Encapsulation:** prefix an attribute with a single underscore (`_x`) to signal it's intended as internal/private.
- **Polymorphism:** different classes can implement the same method name in their own way.
- **Abstraction:** expose only what a user needs, hide the rest.

**Task:** Give two different classes a `describe()` method that behaves differently, and call both through a common loop to see polymorphism in action.

```python
class ToFSensor:
    def __init__(self, resolution):
        self._resolution = resolution   # encapsulated (internal) attribute
        self.__private = "THis is a private attribute"

    def describe(self):
        return f"ToF sensor with resolution {self._resolution}"

class RGBCamera:
    def __init__(self, fps):
        self._fps = fps

    def describe(self):
        return f"RGB camera running at {self._fps} FPS"

devices = [ToFSensor("320x240"), RGBCamera(60)]
for device in devices:
    print(device.describe())
```

## 13. Bonus: Stacks & Queues with `deque`

Python's `list` can act as a stack (LIFO) via `append`/`pop`. For an efficient queue (FIFO), use `collections.deque`.

**Task:** Implement a small `Stack` class, then use `deque` directly to simulate a queue.

```python
class Stack:
    def __init__(self):
        self._items = []

    def push(self, item):
        self._items.append(item)

    def pop(self):
        return self._items.pop()

    def print_stack(self):
        print(self._items)

s = Stack()
for value in [10, 20, 30, 40]:
    s.push(value)
s.print_stack()
print(f"Popped: {s.pop()}")
s.print_stack()
```

```python
task_queue = deque([4])
task_queue.append(6)
task_queue.append(10)
print(f"Queue after adding: {task_queue}")

first_task = task_queue.popleft()
print(f"Processed: {first_task}, remaining: {task_queue}")
```

```python
class Game:
    difficulty = "Easy" # Class variable

g1 = Game()
g2 = Game()

# This looks like it changes the class variable, but it does NOT!
g1.difficulty = "Hard" 

print(g1.difficulty) # "Hard" 
print(g2.difficulty) # "Easy" (The class variable remained unchanged)
print(Game.difficulty) # "Easy"
```
